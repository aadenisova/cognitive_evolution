```python
import pandas as pd

PATH_TO_DIR = "/lustre/fs5/jarv_lab/scratch/adenisova/Inno_2025/conserved_regions/Sackton_et_al"
df = pd.read_csv(
    f"{PATH_TO_DIR}/rer_trees_birds_no_na_by_final_cnees_long.txt", 
    sep = "\t",
    names = ["name", "tree"]
)
```

```python
from ete3 import Tree

df["n_leaves"] = df["tree"].apply(lambda x: len(Tree(x, format=1).get_leaf_names()))
```

```python
df[df["n_leaves"] > 85].hist("n_leaves", bins=100)
```

```python
df[df["n_leaves"] >= 90][[
    "name", "tree"
]].to_csv(
    f"{PATH_TO_DIR}/rer_trees_birds_90_final_cnees_long.txt",
    sep = "\t",
    index = False,
    header = False
)
```
