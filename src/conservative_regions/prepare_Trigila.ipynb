```python
import pandas as pd

PATH = "/lustre/fs5/jarv_lab/scratch/adenisova/Inno_2025/conserved_regions/"

files = {
    "conserved":
        {
            "coding": "11_aves_conserved_coding_elements.bed",
            "noncoding": "12_aves_conserved_noncoding_elements.bed",
        },
    "accelerated":
        {
            "coding": "7_aves_accelerated_coding_elements.bed",
            "noncoding": "8_aves_accelerated_noncoding_elements.bed",
        }
}

df_report = pd.read_csv(
    f"{PATH}/genome_data/GCF_000002315.6_GRCg6a_assembly_report.txt",
    sep = "\t",
)
df_report["seqnames"] = df_report["UCSC-style-name"]
df_report = df_report[["seqnames", "RefSeq-Accn"]]

for elem_type in files.keys():
    tables_for_element = []
    for coding_type in files[elem_type].keys():
        df = pd.read_csv(
            f"{PATH}/Trigila_et_al_2025/selected_elements/{files[elem_type][coding_type]}", 
            sep = "\s+",
        )
        df = df.merge(df_report)
        df["coding_type"] = coding_type 
        tables_for_element.append(df)
    pd.concat(tables_for_element)[[
        "RefSeq-Accn", "start", "end", "coding_type", 
    ]].to_csv(
        f"{PATH}/Trigila_et_al_2025/selected_elements/{elem_type}_elements_merged.bed", 
        sep = "\t", 
        index = False,
        header=False,
    )
```
